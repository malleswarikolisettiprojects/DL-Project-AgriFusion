  <dd style="margin-bottom: 12px; margin-left: 20px; margin-right: 10px;">

  <dt style="font-weight: normal; margin-top: 10px; list-style-type: disc; display: list-item; margin-left: 15px;">Crop Water Requirement Dataset</dt>
  <dd style="margin-bottom: 12px; margin-left: 20px; margin-right: 10px;">
    Collected using FAO (Food and Agriculture Organization) crop coefficient (Kc) values and crop growth stage information.
  </dd>

  <dt style="font-weight: normal; margin-top: 10px; list-style-type: disc; display: list-item; margin-left: 15px;">Target Variable</dt>
  <dd style="margin-bottom: 12px; margin-left: 20px; margin-right: 10px;">
    Daily Crop Water Requirement (mm/day) was calculated using the FAO-56 standard equation.
  </dd>

```python
# Check for outliers in numerical columns using the IQR method
outlier_cols = []
# Find columns with outliers
for col in num_columns:
    Q1 = irrigation_df[col].quantile(0.25)
    Q3 = irrigation_df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = ((irrigation_df[col] < lower) | (irrigation_df[col] > upper)).sum()
    if outliers > 0:
        outlier_cols.append((col, outliers))
# Print outlier counts
print("Columns with Outliers:\n")
for col, count in outlier_cols:
    print(f"{col}: {count} outliers")
# Plot only columns having outliers
n = len(outlier_cols)
cols = 3
rows = math.ceil(n / cols)
plt.figure(figsize=(18, 5 * rows))
for i, (col, count) in enumerate(outlier_cols, 1):
    plt.subplot(rows, cols, i)
    plt.boxplot(irrigation_df[col], vert=False)
    plt.title(f"{col}\n({count} outliers)")
plt.tight_layout()
plt.show()
```

<dt style="font-family:'Times New Roman', Times, serif; font-size:15px; font-weight:bold;">
3.1 Pre-Requisites
</dt>

```python
# Feature Engineering
fe_df = irrigation_df.copy()
fe_df.head(2)
```

```python
cat_cols = fe_df.select_dtypes(include=["object","str"])
num_cols = fe_df.select_dtypes(include=["number"])
display(cat_cols.head(2))
display(num_cols.head(2))
```

```python
# X & Y
y = fe_df["water_requirement_mm_day"]
#Drop columns
drop_cols = ["district","date", "water_requirement_mm_day","latitude","longitude"]
x = fe_df.drop(columns=drop_cols)
```

<ul style="font-family:'Times New Roman', Times, serif; font-size:15px; margin-top:8px;">
<li>Here district, date, water_requirement are droped before storing all the features into x</li>
</ul>

```python
display(y.head(5))
display(x.head(3))
```

<dt style="font-family:'Times New Roman', Times, serif; font-size:15px; font-weight:bold;">
a) Feature Engineering
</dt>

```python
bins = [0, 0.5, 0.9, 1.2, 2.0]
labels = ["low", "medium", "high", "very_high"]
x["kc_band"] = pd.cut(x["kc"], bins=bins, labels=labels)
x = x.drop(columns=["kc"])
```

```python
# Binary Encoding
x["state"] = x["state"].map({"Telangana": 0, "Andhra Pradesh": 1})
x["climate_risk"] = x["climate_risk"].map({"Low":0, "Medium": 1})
```

```python
# Ordinal Encoding
growth_order = {"Initial": 1, "Late-season": 2, "Development": 3, "Mid-season": 4}
x["growth_stage"] = x["growth_stage"].map(growth_order)
```

```python
print(x.columns.tolist())
```

<dt style="font-family:'Times New Roman', Times, serif; font-size:15px; font-weight:bold;">
b) Train - Test Split
</dt>

```python
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
# Columns to One-Hot Encode
one_hot_cols = ["city", "crop", "soil_type", "kc_band"]
encoders = {}
for col in one_hot_cols:
    enc = OneHotEncoder(sparse_output=False, handle_unknown="ignore")
    train_enc = enc.fit_transform(X_train[[col]])
    test_enc = enc.transform(X_test[[col]])
    train_df = pd.DataFrame(train_enc, columns=enc.get_feature_names_out([col]), index=X_train.index)
    test_df = pd.DataFrame(test_enc, columns=enc.get_feature_names_out([col]), index=X_test.index)
    X_train = pd.concat([X_train.drop(columns=[col]), train_df], axis=1)
    X_test = pd.concat([X_test.drop(columns=[col]), test_df], axis=1)
    encoders[col] = enc
```

<dt style="font-family:'Times New Roman', Times, serif; font-size:15px; font-weight:bold;">
c) Model Training
</dt>

```python
import joblib
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
from sklearn.svm import SVR
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from sklearn.metrics import mean_squared_error, r2_score
```

```python
sample_idx = X_train.sample(20000, random_state=42).index
X_train_small, y_train_small = X_train.loc[sample_idx], y_train.loc[sample_idx]
 
models = {
    "Linear Regression": (LinearRegression(), X_train, y_train),
    "KNN": (KNeighborsRegressor(n_jobs=-1), X_train_small, y_train_small),
    "SVR": (SVR(), X_train_small, y_train_small),
    "Decision Tree": (DecisionTreeRegressor(max_depth=4, min_samples_leaf=100, random_state=42), X_train, y_train),
    "Random Forest": (RandomForestRegressor(n_estimators=100, max_depth=4, min_samples_leaf=100,
                                             random_state=42, n_jobs=-1), X_train, y_train),
    "XGBoost": (XGBRegressor(n_estimators=50, max_depth=3, learning_rate=0.05,
                              subsample=0.7, colsample_bytree=0.7, reg_alpha=1, reg_lambda=1,
                              random_state=42, n_jobs=-1, tree_method="hist"), X_train, y_train),
}
 
def adjusted_r2(r2, n, p):
    return 1 - (1 - r2) * (n - 1) / (n - p - 1)
 
def evaluate_model(model, X_tr, y_tr, X_test, y_test):
    model.fit(X_tr, y_tr)
    train_pred, test_pred = model.predict(X_tr), model.predict(X_test)
    train_rmse = np.sqrt(mean_squared_error(y_tr, train_pred))
    test_rmse = np.sqrt(mean_squared_error(y_test, test_pred))
    train_r2, test_r2 = r2_score(y_tr, train_pred), r2_score(y_test, test_pred)
    adj_r2 = adjusted_r2(test_r2, X_test.shape[0], X_test.shape[1])
    if train_r2 < 0.5 and test_r2 < 0.5:
        fit = "Underfit"
    elif abs(train_r2 - test_r2) > 0.20:
        fit = "Overfit"
    else:
        fit = "Good Fit"
    return {"Train RMSE": train_rmse, "Test RMSE": test_rmse, "Train R2": train_r2,
            "Test R2": test_r2, "Adjusted R2": adj_r2, "Fit": fit}, model
 
results, fitted_models = [], {}
for name, (model, X_tr, y_tr) in models.items():
    print(f"Training {name}...")
    evaluation, fitted = evaluate_model(model, X_tr, y_tr, X_test, y_test)
    evaluation["Model"] = name
    results.append(evaluation)
    fitted_models[name] = fitted
 
results_df = pd.DataFrame(results)[["Model", "Train RMSE", "Test RMSE", "Train R2", "Test R2", "Adjusted R2", "Fit"]]
display(results_df)
```

```python
best_name = results_df.loc[results_df["Test R2"].idxmax(), "Model"]
best_model = fitted_models[best_name]
print(f"Best model: {best_name}")
 
joblib.dump(best_model, "best_irrigation_model.pkl")
joblib.dump(encoders, "onehot_encoders.pkl")
```
